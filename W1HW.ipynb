```python
# Basic Math
a, b = 10, 5
sum = a+b
difference = a-b
product = a*b
division = a/b

print(f"the sum of the two numbers {a}, {b} is {sum}")
print("Difference: ", difference)
print("Product: ", product)
print("Division", division)

quotient = 13//4
remainder = 13%4
print(f"when dividing 13/4, the remainder is {remainder}, and the quotent is {quotient}")
```

```python
# Boolian variables and comparisons
x, y = 10, 5
is_greater = x>y
is_equal = x==y
is_not_equal = x!=y

print("is x greater then y?", is_greater)
print("is x equal y?", is_equal)
print("is x not equal to y?", is_not_equal)

name = 'James'
age = 22
is_student = True

print(f"{name}'s age is {age}, is he a student? {is_student}")
```

```python
# Defining and Using a Function

def greet(name):
  return "Hello, " + name + "!"

inputname = input("Enter your name:")
greet(inputname)

def add_numbers(a,b):
  return a+b
a = eval(input("Enter first number: "))
b = eval(input("Enter second number: "))
addval = add_numbers(a,b)
print(f"The result of addition of {a} and {b} is {addval}")
```

```python
# working with strings
message = "Hello, World"
length = len(message)
uppercase = message.upper()
lowercase = message.lower()
substring = message[7:12]

print("orginal message: ", message)
print("length of message: ", length)
print('uppercase', uppercase)
print('lowercase', lowercase)
print('part of the message is:', substring)
```
